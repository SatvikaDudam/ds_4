```python
import numpy as np
import pandas as pd
from scipy.stats import chi2_contingency, chi2
```

```python
# Chi-Square Formula
χ² = Σ [(Observed - Expected)² / Expected]

Where:
- Observed = Actual data
- Expected = Calculated assuming no relationship

This formula measures how much the observed data deviates from expected data.
```

```python
#1. Hypotheses
- Null Hypothesis (H0):  
  There is no significant difference between device type and customer satisfaction.

- Alternative Hypothesis (H1):  
  There is a significant difference between device type and customer satisfaction.
```

```python
# Observed Data(given)
data = np.array([
    [50, 70],   # Very Satisfied
    [80, 100],  # Satisfied
    [60, 90],   # Neutral
    [30, 50],   # Unsatisfied
    [20, 50]    # Very Unsatisfied
])

# Converting to DataFrame
df = pd.DataFrame(data, 
                  columns=["Smart Thermostat", "Smart Light"],
                  index=["Very Satisfied", "Satisfied", "Neutral", "Unsatisfied", "Very Unsatisfied"])
df
```

```python
#2. Chi-square statistics
chi2_stat, p_value, dof, expected = chi2_contingency(data)

print("Chi-Square Statistic:", chi2_stat)
print("P-value:", p_value)
print("Degrees of Freedom:", dof)
```

```python
#expected frequencies
expected_df = pd.DataFrame(expected, 
                           columns=["Smart Thermostat", "Smart Light"],
                           index=df.index)

print("Expected Frequencies:\n")
expected_df
```

```python
#3. Critical Value
alpha = 0.05

critical_value = chi2.ppf(1 - alpha, dof)
print("Critical Value:", critical_value)
```

```python
#4. Decision
if chi2_stat > critical_value:
    print("Reject Null Hypothesis (H0)")
else:
    print("Fail to Reject Null Hypothesis (H0)") #This means H0 is accepted and H1 is rejected.
```

```python
Interpretation
Ho is accepted and H1 is rejected.
Thus there is no significant difference between device type and customer satisfaction to
suggest a relationship between them.
Any differences observed in satisfaction levels may be due to random variation ranther 
than actual effect.
```

```python
Conclusion
Chi square test is the one of the simplest method mostly used in non parametric test in statistical world.
Customer satisfaction is independent of the type of smart device purchased.
```
