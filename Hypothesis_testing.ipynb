```python
import numpy as np
from scipy.stats import norm
```

```python
Z-Test:
A Z-test formula determines if there is a significant difference between sample(x̄) and population means(μ) when sample size is large (n>30)
or variance is known, using the standard normal distribution.
```

```python
#Given Data:
Sample_mean = 3050 # x̄ (sample mean)
n=25 #sample size
X_mean=600 #mean units produced
sigma_X=25 #standard deviation of units

#cost model: W=1000+5X
#so theroritical mean :
mu = 1000+5*X_mean # mu=1000 + 5(600)

#Standard deviation of W:
sigma = 5*sigma_X

print("Theoretical Mean(mu):",mu)
print("Standard Deviation(sigma):",sigma)
```

```python
#Hypothesis:
#H0: μ = 4000 
#H1: μ > 4000 
print("H0:μ=4000") #(There is no significant difference between μ and x̄ )
print("H1:μ>4000") #(There is significant difference between μ and x̄ )
```

```python
#Test Statistics (Z):
# Formula:
# Z = (x̄ - μ) / (σ / sqrt(n))

z = (Sample_mean - mu) / (sigma / np.sqrt(n))

print("Test Statistic (Z):", z)
```

```python
#Critical Value:
#Alpha = 0.05 (one-tailed test)
alpha = 0.05

# For right-tailed test
critical_value = norm.ppf(1 - alpha)

print("Critical Value (Zα):", critical_value)
```

```python
#Decision:
if z > critical_value:
    print("Reject Null Hypothesis (H0)")
else:
    print("Fail to Reject Null Hypothesis (H0)")
```

```python
Interpretation:
H0 is accepted and H1 is rejected.
So, there is no significant difference between μ and x̄ .
```

```python
Conclusion:
There is not enough evidence to support the claim that costs are higher than the model.
In fact, the sample mean is significantly lower than the expected cost, suggesting the model may still hold or costs may even be lower.
```
